```python
def convertToNumber(thing):
    if thing == "one": return 1
    if thing == "two": return 2
    if thing == "three": return 3
    if thing == "four": return 4
    if thing == "five": return 5
    if thing == "six": return 6
    if thing == "seven": return 7
    if thing == "eight": return 8
    if thing == "nine": return 9
    if thing == "ten": return 10
    return thing
```
